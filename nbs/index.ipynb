# craft_dev

`craft_dev` is an `nbdev` toolkit for authoring, building, testing, and deploying `solve-it-crafts` customizations for Solveit.

## Notebook layout

- `nbs/00_auth.ipynb` → `craft_dev/auth.py` — Chrome/Vivaldi OAuth login and cookie storage (`.auth/`, gitignored)
- `nbs/01_cdp.ipynb` → `craft_dev/cdp.py` — `open_solveit()` CDP `Fetch` local asset interception, `inspect_dialog()`, `smoke_dialog()`
- `nbs/02_cli.ipynb` → `craft_dev/cli.py` — `craft-cdp` CLI entry point
- `nbs/03_build.ipynb` → `craft_dev/build.py` — `#| export_js` / `export_js_iife` / `export_css` / `export_ft` compiler and `build_dist()`
- `nbs/04_tunnel.ipynb` → `craft_dev/tunnel.py` — private Tailscale + rustygate tunnel and atomic `deploy_dist()`
- `nbs/05_chrome_ext.ipynb` → `craft_dev/chrome_ext.py` — Manifest V3 Chrome/Vivaldi extension (`chrome-ext/`) that stubs Solveit `modern-monaco` and `tm-grammars` requests before load
- `nbs/craft/00_architecture.ipynb` — UI architecture map, dependency layers, and Solveit DOM/runtime contracts
- `nbs/craft/01_theme_and_tokens.ipynb` .. `19_bug_capture.ipynb` — feature-owned CRAFT UI notebooks ordered by load/cascade layer (`01_theme_and_tokens`, `02_hotkey_builtins`, `03_hotkey_manager`, `04_cm_core`, `05_dialog_chrome`, `06_file_tree`, `07_cell_presentation`, `08_tables`, `09_cell_insertion`, `10_code_editor_c2`, `11_markdown_editor`, `12_mobile`, `13_context_control`, `14_font_settings`, `15_structural_ops`, `16_navigation_reference`, `17_cell_history`, `18_tts_reader`, `19_bug_capture`)

## Quick start

1. Log in once via Chrome or Vivaldi (Google OAuth works natively):

```bash
uv run craft-cdp auth --url "https://your-instance.solve.it.com/dialog_?name=CRAFT"
```

2. Build `dist/` from `nbs/craft/*.ipynb` and run checks:

```bash
uv run craft-cdp build              # compile nbs/craft/ -> dist/ and run node --check (+ tsc if installed)
uv run craft-cdp build-ext          # generate unpacked MV3 Chrome/Vivaldi extension in chrome-ext/
uv run craft-cdp auth-status        # verify saved session
uv run craft-cdp inspect --assets-dir dist --screenshot .craft-debug/dialog.png
uv run craft-cdp smoke --assets-dir dist
uv run nbdev-test                   # offline notebook tests
uv run nbdev-test --flags live      # offline + live fastcdp UI tests
```

3. Deploy to your live Solveit instance or sync shipped assets back to the repo root for `setup.sh`:

```bash
uv run craft-cdp deploy             # build dist/ and deploy over private rustygate tunnel
uv run craft-cdp build --sync       # build dist/, sync shipped assets to repo root, and rebuild chrome-ext/
```

End-user installations via `bash setup.sh` copy only the runtime UI assets (`CRAFT.css`, `CRAFT.js`, `.craft-assets/*`) and never install `nbs/`, `craft_dev/`, or `dev/` into Solveit.

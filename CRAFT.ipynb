Proyecto: Aplicación inteligente para el aprendizaje de euskera orientada a hablantes de español

## Objetivo
Desarrollar una aplicación inteligente para el aprendizaje de euskera (vasco) orientada a hablantes de español, capaz de adaptar los contenidos y actividades al conocimiento y progreso individual de cada estudiante mediante técnicas de Procesamiento de Lenguaje Natural.

## Introducción general al tema
El aprendizaje del euskera plantea barreras complejas para los hispanohablantes debido a su marcada distancia lingüística, especialmente en el plano morfosintáctico y en el sistema verbal. Aunque históricamente minorizada, la lengua atraviesa un proceso activo de revitalización institucional que ha impulsado su estudio tanto entre residentes locales como en una amplia comunidad en la diáspora. Sin embargo, el acceso a herramientas pedagógicas flexibles y eficaces sigue siendo limitado, lo que dificulta atender esta demanda.
Como ocurre con cualquier lengua, el aprendizaje no sigue una progresión homogénea: un estudiante puede dominar cierto léxico, tropezar con estructuras gramaticales específicas o asimilar unos conceptos con mayor facilidad que otros. Ante estas asimetrías, los itinerarios lineales y estandarizados resultan insuficientes. La solución propuesta registrará el historial del usuario, modelará su nivel de competencia en distintas áreas lingüísticas y evaluará los recursos disponibles para prescribir el contenido óptimo en cada etapa. Así, el sistema seleccionará actividades y textos accesibles que, al mismo tiempo, refuercen nociones débiles e introduzcan gradualmente nuevo vocabulario y estructuras. Para sostener esta adaptabilidad, se emplearán técnicas de Procesamiento del Lenguaje Natural (PLN) encargadas de analizar tanto el material didáctico como las necesidades pedagógicas de cada estudiante.

## Descripción detallada
El sistema se concibe como un entorno adaptativo enfocado en la práctica deliberada y la producción escrita para niveles iniciales a intermedios (A1–B1). Su diseño pedagógico combina sesiones breves con retroalimentación inmediata, integrando principios de recuperación activa y repetición espaciada para maximizar la retención. Para garantizar una progresión controlada, la fase inicial prioriza la interacción puramente textual sobre un corpus léxico-gramatical acotado, reservando la evaluación fonética y la interacción oral para etapas posteriores de desarrollo.
El sistema se articula en la arquitectura modular de cuatro bloques ilustrada en la Figura 1. En primer lugar, la capa de interfaz gestiona la interacción directa del usuario y la entrega dinámica de actividades, mientras que el módulo de registro y telemetría captura eventos, latencias y patrones de respuesta. Esta información alimenta al núcleo adaptativo, encargado de contrastar el modelo cognitivo del estudiante con el motor de prescripción para secuenciar el contenido óptimo. Por último, el módulo de conocimiento lingüístico y pipeline de PLN procesa las respuestas abiertas en euskera, validando formulaciones alternativas admisibles y diagnosticando errores morfosintácticos específicos.

## Disponibilidad de datos
Existen corpus y recursos de Procesamiento de Lenguaje Natural disponibles para euskera, incluyendo corpus monolingües y paralelos vasco-español, además de herramientas de análisis morfológico y sintáctico. Estos recursos podrán utilizarse como fuente para obtener ejemplos, analizar vocabulario y estructuras y construir una base inicial de contenido. También será necesario desarrollar una colección propia y acotada de materiales educativos, incluyendo vocabulario, frases, textos, ejercicios y sus conceptos asociados.
